```python
import os, time
from pathlib import Path

FOLDER = Path("/content/messy")
OLD_TIME = time.time() - 3600  # pretend the files are an hour old
SAMPLES = {
    "IMG_2231.txt": "Maths unit 2 notes: integration by parts, examples and practice questions.",
    "final_FINAL2.txt": "Electricity bill for October. Amount due Rs 1,240 by 12 Nov.",
    "doc (3).txt": "Meeting notes: client wants the report draft by Friday, send slides too.",
    "new.txt": "Hinglish recipe notes: aloo paratha banane ke liye aata gunthna hai.",
}

FOLDER.mkdir(exist_ok=True)
for name, text in SAMPLES.items():
    path = FOLDER / name
    path.write_text(text, encoding="utf-8")
    os.utime(path, (OLD_TIME, OLD_TIME))
print("Demo files ready:", sorted(os.listdir(FOLDER)))
```

```python
MIN_FILE_AGE_SECONDS = 60


def list_files(folder):
    now = time.time()
    found = []
    for path in sorted(folder.iterdir()):
        try:
            if not path.is_file() or path.name.startswith("."):
                continue
            if now - path.stat().st_mtime < MIN_FILE_AGE_SECONDS:
                continue
        except OSError:
            continue
        found.append(path)
    return found


if not FOLDER.is_dir():
    print(f"Folder not found: {FOLDER}")
else:
    files = list_files(FOLDER)
    if not files:
        print("Nothing to tidy.")
    for file in files:
        print(file.name)
```

```python
import json
import urllib.error
import urllib.request

OLLAMA_URL = "http://localhost:11434/api/generate"
MODEL_NAME = "gemma3:4b"
REQUEST_TIMEOUT_SECONDS = 180
CATEGORIES = ("Study", "Work", "Money", "Photos", "Installers", "Documents", "Other")
MAX_PREVIEW_CHARS = 500


def read_preview(path):
    """Read the first few characters of a text file."""
    try:
        with path.open("r", encoding="utf-8", errors="ignore") as handle:
            return " ".join(handle.read(MAX_PREVIEW_CHARS).split())
    except OSError:
        return ""


def build_prompt(file_name, preview):
    return (
        "You are a careful file organizer. Reply with JSON only.\n"
        "Choose the best category from this list: " + ", ".join(CATEGORIES) + ".\n"
        "Also suggest a short, clear file name of 2 to 6 words. "
        "Do not include the file extension or special characters.\n"
        "File name: " + file_name + "\n"
        "Start of the file (may be empty): " + (preview or "(not available)") + "\n"
        'Reply exactly like this: {"category": "...", "new_name": "..."}'
    )


def ask_model(prompt):
    """Send the prompt to Ollama. Returns a dict, or None if the answer was unusable."""
    body = json.dumps(
        {"model": MODEL_NAME, "prompt": prompt, "stream": False, "format": "json"}
    ).encode("utf-8")
    request = urllib.request.Request(
        OLLAMA_URL, data=body, headers={"Content-Type": "application/json"}
    )
    try:
        with urllib.request.urlopen(request, timeout=REQUEST_TIMEOUT_SECONDS) as reply:
            payload = json.loads(reply.read().decode("utf-8"))
    except urllib.error.HTTPError as error:
        if error.code == 404:
            raise RuntimeError(f"Model not installed. Run: !ollama pull {MODEL_NAME}") from error
        raise RuntimeError(f"Ollama returned an error (code {error.code}).") from error
    except (urllib.error.URLError, ConnectionError) as error:
        raise RuntimeError("Cannot reach Ollama. Run Cell 2 again.") from error
    except TimeoutError:
        print("The AI was too slow, try again.")
        return None

    try:
        answer = json.loads(payload.get("response", ""))
    except (json.JSONDecodeError, TypeError, AttributeError):
        return None
    return answer if isinstance(answer, dict) else None


# ---- Try it on the first file ----
if not files:
    print("No files found. Run Cells 3 and 4 first.")
else:
    first = files[0]
    print("Asking the AI about:", first.name)
    answer = ask_model(build_prompt(first.name, read_preview(first)))
    print("AI says:", answer)
```

```python
import re

FALLBACK_CATEGORY = "Other"
MAX_NAME_LENGTH = 60
BAD_NAME_CHARS = re.compile(r'[<>:"/\\|?*\x00-\x1f]')
RESERVED_WINDOWS_NAMES = {"CON", "PRN", "AUX", "NUL"} | {
    f"{prefix}{n}" for prefix in ("COM", "LPT") for n in range(1, 10)
}


def clean_category(value):
    """Return a category from our list, or 'Other' if the AI made one up."""
    if isinstance(value, str):
        for category in CATEGORIES:
            if category.lower() == value.strip().lower():
                return category
    return FALLBACK_CATEGORY


def clean_stem(value, original):
    """Turn the AI's suggested name into a safe Windows file name (no extension)."""
    if not isinstance(value, str):
        return original.stem
    text = value.strip()
    if original.suffix and text.lower().endswith(original.suffix.lower()):
        text = text[: -len(original.suffix)]
    text = BAD_NAME_CHARS.sub(" ", text)
    text = " ".join(text.split()).strip(" .")[:MAX_NAME_LENGTH].strip(" .")
    if not text or text.upper() in RESERVED_WINDOWS_NAMES:
        return original.stem
    return text


# ---- Test it with bad answers on purpose ----
sample = files[0]
print(clean_category("study"))          # wrong case
print(clean_category("Hacking"))        # not on our list
print(clean_category(None))             # not even text
print(clean_stem("Maths: Unit/2.txt", sample))  # illegal characters
print(clean_stem("CON", sample))        # reserved Windows name
print(clean_stem(123, sample))          # not text
```

```python
OUTPUT_FOLDER_NAME = "Tidied"


def unique_destination(destination, taken):
    """Add (1), (2)... if a name is already used. Windows ignores upper/lower case."""
    candidate = destination
    counter = 1
    while candidate.exists() or str(candidate).lower() in taken:
        candidate = destination.with_name(f"{destination.stem} ({counter}){destination.suffix}")
        counter += 1
    taken.add(str(candidate).lower())
    return candidate


def make_plan(folder, file_list):
    """Ask the AI about every file and return a list of planned moves."""
    output_root = folder / OUTPUT_FOLDER_NAME
    taken = set()
    plan = []
    total = len(file_list)
    for index, path in enumerate(file_list, start=1):
        print(f"[{index}/{total}] Sloth is thinking about: {path.name}")
        answer = ask_model(build_prompt(path.name, read_preview(path))) or {}
        category = clean_category(answer.get("category"))
        new_name = clean_stem(answer.get("new_name"), path) + path.suffix
        destination = unique_destination(output_root / category / new_name, taken)
        plan.append({"source": path, "destination": destination})
    return plan


def show_plan(plan, folder):
    print("\nHere is the plan:\n")
    for item in plan:
        relative = item["destination"].relative_to(folder)
        print(f"  {item['source'].name}\n     -> {relative}")


# ---- Run it ----
if not files:
    print("No files found. Run Cells 3 and 4 first.")
else:
    try:
        plan = make_plan(FOLDER, files)
        show_plan(plan, FOLDER)
        print("\nNothing was moved yet.")
    except RuntimeError as error:
        print(f"\nProblem: {error}")
```

```python
CATEGORIES = ("Study", "Work", "Money", "Food", "Photos", "Installers", "Documents", "Other")


def clean_stem(value, original):
    """Turn the AI's suggested name into a safe Windows file name (no extension)."""
    if not isinstance(value, str):
        return original.stem
    text = value.strip()
    if original.suffix and text.lower().endswith(original.suffix.lower()):
        text = text[: -len(original.suffix)]
    text = BAD_NAME_CHARS.sub(" ", text.replace("_", " "))
    text = " ".join(text.split()).strip(" .")[:MAX_NAME_LENGTH].strip(" .")
    if not text or text.upper() in RESERVED_WINDOWS_NAMES:
        return original.stem
    return text[0].upper() + text[1:]
```

```python
import json
import shutil
from pathlib import Path

LOG_FILE = Path("/content/sloth_undo_log.json")


def apply_plan(plan):
    """Move the files and save a log of every move so it can be undone."""
    done = []
    for item in plan:
        source, destination = item["source"], item["destination"]
        try:
            destination.parent.mkdir(parents=True, exist_ok=True)
            shutil.move(str(source), str(destination))
        except OSError as error:
            print(f"  Could not move {source.name}: {error}")
            continue
        done.append({"from": str(source), "to": str(destination)})
    try:
        LOG_FILE.write_text(json.dumps(done, indent=2, ensure_ascii=False), encoding="utf-8")
    except OSError as error:
        print(f"Warning: could not save the undo log ({error}). Undo will not work.")
    return len(done)


if not plan:
    print("The plan is empty. Run Cell 7 first.")
else:
    moved = apply_plan(plan)
    print(f"Done. Moved {moved} of {len(plan)} file(s).\n")
    for path in sorted(FOLDER.rglob("*")):
        print(" ", path.relative_to(FOLDER))
```

```python
def undo_last_run():
    """Put the files from the last apply back where they were."""
    if not LOG_FILE.exists():
        print("Nothing to undo: no earlier run was found.")
        return
    try:
        entries = json.loads(LOG_FILE.read_text(encoding="utf-8"))
        if not isinstance(entries, list):
            raise ValueError("log is not a list")
    except (OSError, ValueError) as error:
        print(f"The undo log is unreadable ({error}).")
        return

    restored = 0
    for entry in reversed(entries):
        try:
            moved, original = Path(entry["to"]), Path(entry["from"])
        except (KeyError, TypeError):
            continue
        if not moved.exists():
            print(f"  Skipping {moved.name}: it is no longer there.")
            continue
        if original.exists():
            print(f"  Skipping {moved.name}: a file with the old name already exists.")
            continue
        try:
            shutil.move(str(moved), str(original))
            restored += 1
            for empty_folder in (moved.parent, moved.parent.parent):
                try:
                    empty_folder.rmdir()  # only works if the folder is empty
                except OSError:
                    break
        except OSError as error:
            print(f"  Could not restore {moved.name}: {error}")
    print(f"Put back {restored} file(s). Back to the messy life!")


undo_last_run()
for path in sorted(FOLDER.rglob("*")):
    print(" ", path.relative_to(FOLDER))
```
